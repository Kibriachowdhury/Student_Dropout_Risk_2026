```python
import pandas as pd

# Importing pandas and loading the compressed dataset from Google Drive
url = 'https://drive.google.com/uc?export=download&id=1P6eZPCg1FwpXnbq2Yt52RTdvDLk4PSD9'
df = pd.read_csv(url, compression='gzip')

df.head()
```

```python
# Displaying all columns without truncation to inspect the first few rows
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
df.head()
```

## 1. Extract all students that study `Online`, have an `attendance_rate` smaller than 50, and have more than 5 `missed_deadlines`. Store them in a dataframe called `at_risk`.

```python
# Filtering online students with attendance under 50% and over 5 missed deadlines
online_students = df['study_mode'] == 'Online'
low_attendance = df['attendance_rate'] < 50
many_missed_deadlines = df['missed_deadlines'] > 5

at_risk = df[online_students & low_attendance & many_missed_deadlines].copy()
at_risk.head()
```

```python
## 2. Then compute, for each `course_type`, how many of those students there are and what fraction of them has dropped out. Store the result in a dataframe called `at_risk_by_course`, with columns students and dropout_rate, sorted by decreasing dropout_rate.
```

```python
# Grouping at-risk students by course type and sorting by highest dropout rate
students = at_risk.groupby('course_type')['person_id'].count()
dropout_rate = at_risk.groupby('course_type')['dropout'].mean()

at_risk_by_course = pd.DataFrame({
    'students': students,
    'dropout_rate': dropout_rate,
}).sort_values('dropout_rate', ascending=False)

at_risk_by_course
```

## 3. Add to the dataframe the following columns:

1. `pass_rate`: the ratio between `units_passed` and `units_enrolled`;
2. `gpa_change`: the difference between `gpa_semester_2` and `gpa_semester_1`;
3. `gpa_trend`: a string that is `"improving"` if `gpa_change` is larger than 0.5, `"declining"` if `gpa_change` is smaller than -0.5, and `"stable"` otherwise.

```python
# 1. Calculating pass_rate as units_passed divided by units_enrolled, filling missing values with 0.0
df["pass_rate"] = (df["units_passed"] / df["units_enrolled"]).fillna(0.0)

# 2. Computing gpa_change as the difference between gpa_semester_2 and gpa_semester_1
df["gpa_change"] = df["gpa_semester_2"] - df["gpa_semester_1"]

# 3. Assigning "stable" by default and updating gpa_trend based on gpa_change thresholds
df["gpa_trend"] = "stable"
df.loc[df["gpa_change"] > 0.5, "gpa_trend"] = "improving"
df.loc[df["gpa_change"] < -0.5, "gpa_trend"] = "declining"

df[
    [
        "units_enrolled",
        "units_passed",
        "pass_rate",
        "gpa_semester_1",
        "gpa_semester_2",
        "gpa_change",
        "gpa_trend",
    ]
].head()
```

### 4. Verify that, for every student, `units_passed + units_failed` equals `units_enrolled`; if not, report the rows violating this condition.

```python
# 1. Finding rows where units_passed + units_failed does not equal units_enrolled
mismatches = df[df["units_passed"] + df["units_failed"] != df["units_enrolled"]]

# 2. Checking how many rows violate the rule and reporting them if any exist
print("Violating rows:", len(mismatches))

if len(mismatches) > 0:
    display(mismatches[["person_id", "units_enrolled", "units_passed", "units_failed"]])
else:
    print("Every student satisfies units_passed + units_failed == units_enrolled.")
```

### 5. Compute the dropout rate for each value of `gpa_trend`, and the `person_id` of the 10 students with the largest GPA drop.

```python
# 1. Computing the dropout rate for each gpa_trend category
dropout_rate = df.groupby("gpa_trend")["dropout"].mean()
print("Dropout rate by gpa_trend:")
print(dropout_rate)

# 2. Sorting by gpa_change ascending (largest drops first) and selecting the top 10 person_ids
top_10_drops = df.sort_values(by="gpa_change", ascending=True).head(10)["person_id"]
print("\nTop 10 students with the largest GPA drop (person_id):")
print(top_10_drops)
```

### 6. Partition the students into the following groups, according to `weekly_work_hours`: `0`, `1-10`, `11-20`, `21-30`, `more than 30` (all intervals include both extremes).

```python
# 1. Defining a helper function to map weekly_work_hours to its group
def get_work_group(hours):
    if hours == 0:
        return "0"
    elif 1 <= hours <= 10:
        return "1-10"
    elif 11 <= hours <= 20:
        return "11-20"
    elif 21 <= hours <= 30:
        return "21-30"
    else:
        return "more than 30"

# 2. Applying the function to create the work_hours_group column
df["work_hours_group"] = df["weekly_work_hours"].apply(get_work_group)

# 3. Checking the count in each group
print(df["work_hours_group"].value_counts())
```
