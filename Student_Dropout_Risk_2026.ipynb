```python
import pandas as pd

# Importing pandas and loading the compressed dataset from Google Drive
url = 'https://drive.google.com/uc?export=download&id=1P6eZPCg1FwpXnbq2Yt52RTdvDLk4PSD9'
df = pd.read_csv(url, compression='gzip')

df.head()
```

```python
# Displaying all columns without truncation to inspect the first few rows
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
df.head()
```

## 1. Extract all students that study `Online`, have an `attendance_rate` smaller than 50, and have more than 5 `missed_deadlines`. Store them in a dataframe called `at_risk`.

```python
# Filtering online students with attendance under 50% and over 5 missed deadlines
online_students = df['study_mode'] == 'Online'
low_attendance = df['attendance_rate'] < 50
many_missed_deadlines = df['missed_deadlines'] > 5

at_risk = df[online_students & low_attendance & many_missed_deadlines].copy()
at_risk.head()
```

```python
## 2. Then compute, for each `course_type`, how many of those students there are and what fraction of them has dropped out. Store the result in a dataframe called `at_risk_by_course`, with columns students and dropout_rate, sorted by decreasing dropout_rate.
```

```python
# Grouping at-risk students by course type and sorting by highest dropout rate
students = at_risk.groupby('course_type')['person_id'].count()
dropout_rate = at_risk.groupby('course_type')['dropout'].mean()

at_risk_by_course = pd.DataFrame({
    'students': students,
    'dropout_rate': dropout_rate,
}).sort_values('dropout_rate', ascending=False)

at_risk_by_course
```

## 3. Add to the dataframe the following columns:

1. `pass_rate`: the ratio between `units_passed` and `units_enrolled`;
2. `gpa_change`: the difference between `gpa_semester_2` and `gpa_semester_1`;
3. `gpa_trend`: a string that is `"improving"` if `gpa_change` is larger than 0.5, `"declining"` if `gpa_change` is smaller than -0.5, and `"stable"` otherwise.

```python
# 1. Calculating pass_rate as units_passed divided by units_enrolled, filling missing values with 0.0
df["pass_rate"] = (df["units_passed"] / df["units_enrolled"]).fillna(0.0)

# 2. Computing gpa_change as the difference between gpa_semester_2 and gpa_semester_1
df["gpa_change"] = df["gpa_semester_2"] - df["gpa_semester_1"]

# 3. Assigning "stable" by default and updating gpa_trend based on gpa_change thresholds
df["gpa_trend"] = "stable"
df.loc[df["gpa_change"] > 0.5, "gpa_trend"] = "improving"
df.loc[df["gpa_change"] < -0.5, "gpa_trend"] = "declining"

df[
    [
        "units_enrolled",
        "units_passed",
        "pass_rate",
        "gpa_semester_1",
        "gpa_semester_2",
        "gpa_change",
        "gpa_trend",
    ]
].head()
```

### 4. Verify that, for every student, `units_passed + units_failed` equals `units_enrolled`; if not, report the rows violating this condition.

```python
# 1. Finding rows where units_passed + units_failed does not equal units_enrolled
mismatches = df[df["units_passed"] + df["units_failed"] != df["units_enrolled"]]

# 2. Checking how many rows violate the rule and reporting them if any exist
print("Violating rows:", len(mismatches))

if len(mismatches) > 0:
    display(mismatches[["person_id", "units_enrolled", "units_passed", "units_failed"]])
else:
    print("Every student satisfies units_passed + units_failed == units_enrolled.")
```

### 5. Compute the dropout rate for each value of `gpa_trend`, and the `person_id` of the 10 students with the largest GPA drop.

```python
# 1. Computing the dropout rate for each gpa_trend category
dropout_rate = df.groupby("gpa_trend")["dropout"].mean()
print("Dropout rate by gpa_trend:")
print(dropout_rate)

# 2. Sorting by gpa_change ascending (largest drops first) and selecting the top 10 person_ids
top_10_drops = df.sort_values(by="gpa_change", ascending=True).head(10)["person_id"]
print("\nTop 10 students with the largest GPA drop (person_id):")
print(top_10_drops)
```

### 6. Partition the students into the following groups, according to `weekly_work_hours`: `0`, `1-10`, `11-20`, `21-30`, `more than 30` (all intervals include both extremes).

```python
# 1. Defining a helper function to map weekly_work_hours to its group
def get_work_group(hours):
    if hours == 0:
        return "0"
    elif 1 <= hours <= 10:
        return "1-10"
    elif 11 <= hours <= 20:
        return "11-20"
    elif 21 <= hours <= 30:
        return "21-30"
    else:
        return "more than 30"

# 2. Applying the function to create the work_hours_group column
df["work_hours_group"] = df["weekly_work_hours"].apply(get_work_group)

# 3. Checking the count in each group
print(df["work_hours_group"].value_counts())
```

### 7. Build a risk score by summing:

1. 2 points if the tuition is not up to date;
2. 1 point if `stress_level` is at least 8;
3. 1 point if `sleep_hours` is smaller than 5;
4. 1 point if the student has a part-time job with more than 20 weekly work hours;
5. 2 points if `attendance_rate` is smaller than 60;
6. 1 point if `missed_deadlines` is larger than 3.
For each value of the risk score, compute the number of students and the dropout rate.

```python
# 1. Defining a helper function to calculate risk_score row by row
def calculate_risk_score(row):
    score = 0
    if not row["tuition_up_to_date"]:
        score += 2
    if row["stress_level"] >= 8:
        score += 1
    if row["sleep_hours"] < 5:
        score += 1
    if row["part_time_job"] and row["weekly_work_hours"] > 20:
        score += 1
    if row["attendance_rate"] < 60:
        score += 2
    if row["missed_deadlines"] > 3:
        score += 1
    return score

# 2. Creating the risk_score column across all rows
df["risk_score"] = df.apply(calculate_risk_score, axis=1)

# 3. Computing student count and dropout rate for each risk_score value
summary = df.groupby("risk_score")["dropout"].agg(["count", "mean"])
print(summary)
```

### 8. Read the GDP per capita dataset published by Our World in Data directly from the URL

`https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true`
and store it in a dataframe called `gdp`. The column `ny_gdp_pcap_pp_kd` contains the GDP per capita.

Clean the dataset, removing the rows that do not correspond to a single country (e.g. `World`, `High-income countries`). For each country keep only the row with the most recent `year`.

```python
import io
import requests
# 1. Fetching the CSV with a standard User-Agent header to avoid 403 Forbidden
url = "https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true"
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
response.raise_for_status()

# 2. Loading the downloaded CSV text into a pandas DataFrame
gdp = pd.read_csv(io.StringIO(response.text))

# 3. Standardizing column names to lowercase
gdp.columns = gdp.columns.str.lower()

# 4. Removing non-country aggregates and missing GDP values
gdp = gdp[
    gdp["code"].notna()
    & (~gdp["code"].astype(str).str.startswith("OWID_"))
    & gdp["ny_gdp_pcap_pp_kd"].notna()
].copy()

# 5. Keeping only the most recent year row per country
gdp = (
    gdp.sort_values("year")
    .drop_duplicates(subset=["entity"], keep="last")
    .reset_index(drop=True)
)

# 6. Checking the resulting DataFrame
print("Cleaned shape:", gdp.shape)
display(gdp[["entity", "code", "year", "ny_gdp_pcap_pp_kd"]].head(10))
```

### 9. Join the GPD data with the students data. Compute the dropout rate for each continent.

```python
# 1. Joining students data (df) with GDP data (gdp) on nationality and entity
merged_df = df.merge(gdp, left_on="nationality", right_on="entity", how="left")

# 2. Mapping each country to its continent with a simple function
def get_continent(country):
    if country in ["USA", "Canada", "Mexico"]:
        return "North America"
    elif country in ["Brazil", "Argentina", "Chile", "Colombia", "Peru"]:
        return "South America"
    elif country in ["UK", "United Kingdom", "Germany", "France", "Italy", "Spain", "Portugal", "Netherlands", "Poland", "Sweden", "Norway", "Denmark", "Switzerland", "Austria", "Belgium", "Ireland", "Greece", "Romania", "Ukraine"]:
        return "Europe"
    elif country in ["India", "China", "Bangladesh", "Pakistan", "Japan", "South Korea", "Indonesia", "Vietnam", "Philippines", "Thailand", "Malaysia", "Turkey", "Saudi Arabia", "UAE", "United Arab Emirates", "Israel"]:
        return "Asia"
    elif country in ["Nigeria", "Egypt", "South Africa", "Kenya", "Ethiopia", "Ghana", "Morocco", "Algeria", "Tunisia"]:
        return "Africa"
    elif country in ["Australia", "New Zealand"]:
        return "Oceania"
    else:
        return "Other"

# 3. Creating continent column and calculating dropout rate per continent
merged_df["continent"] = merged_df["nationality"].apply(get_continent)
print(merged_df.groupby("continent")["dropout"].mean())
```
